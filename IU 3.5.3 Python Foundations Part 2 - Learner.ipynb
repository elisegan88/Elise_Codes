## 1.1.2 Indexing

**How do we get specific items from our list?**

Python uses **index numbers** (positions) to access items - just like row numbers in Excel, except they start at 0!

Let's use days of the week as an example - Monday is first, Tuesday is second, etc.



**Understanding positions:**

- First day: `days[0]` → "Monday" (position 0)
- Second day: `days[1]` → "Tuesday" (position 1)
- Last day: `days[-1]` → "Sunday" (negative index counts from the end)

````python
# Create a list of days of the week
days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

# Count how many days
print(f"There are {len(days)} days in the week.")

# Get the first day (position 0)
print(f"The first day is {days[0]}")
````

## 1.1.3 Slicing

**Getting Multiple Items at Once**

What if you want several days from your list? Use **slicing**: `list[start:stop]`

**Key rule:** The start position is **included**, but the stop position is **excluded** (it stops before that number)



**Examples:**
- `days[0:3]` → Gets Monday, Tuesday, Wednesday (positions 0, 1, 2 - stops **before** position 3)
- `days[1:4]` → Gets Tuesday, Wednesday, Thursday (positions 1, 2, 3)
- `days[:3]` → Gets first 3 days (Monday through Wednesday)

- `days[5:]` → Gets weekend (Saturday, Sunday - from position 5 onwards)- `days[-2:]` → Gets the last 2 days (Saturday, Sunday)


````python
# Get weekdays (Monday through Friday)
print("Weekdays:", days[0:5])

# Get middle of the week (Tuesday to Thursday)
print("Midweek:", days[1:4])

# Get the weekend (Saturday and Sunday)
print("Weekend:", days[5:])
````

## 1.2 Tuples

**Think of Excel:**
- **List** = A column of similar data (eg list of contact names, list of tools, list of days)
- **Tuple** = A single row representing one complete record, and every row has similar structure

**The Key Difference:**

**Lists** store **similar items** you want to process together:
```python
contacts = ['Alice', 'Bob', 'Charlie']  # All names - same type of data
days = ['Monday', 'Tuesday', 'Wednesday']  # All days - same type of data
```

**Tuples** store **one complete record** with different fields that belong together:
```python
alice_info = ('Alice', '+65-9876-5432', 'alice@email.com')
#             name     phone            email
```

**Why are Tuples locked?**

Because they represent a **data structure with meaning**:
- Position 0 = Name (always)
- Position 1 = Phone (always)
- Position 2 = Email (always)

Changing the structure would break the meaning! The program would not know how to process a piece of data at position 3 for example

**Syntax:**
- **Lists use square brackets:** `['Alice', 'Bob', 'Charlie']`
- **Tuples use parentheses:** `('Alice', '+65-9876-5432', 'alice@email.com')`

````python
# TUPLE = One complete contact record (name, phone, email)
alice_info = ('Alice', '+65-9876-5432', 'alice@email.com')

# Print the entire tuple and its type
print("Alice's contact info:", alice_info)
print("Type:", type(alice_info))
print()

# Access individual fields by position
print("Name:", alice_info[0])
print("Phone:", alice_info[1])
print("Email:", alice_info[2])
````

**What happens if we try to change a tuple?**

Unlike lists, tuples are **locked** - you cannot change items once created.

````python
# Try to change Alice's name by uncommenting the line below, then run the cell - this will cause an ERROR!
# alice_info[0] = 'Alicia'
````

**Comparing Lists vs Tuples**

Let's see the difference side by side:

````python
# LIST - Flexible (can change)
weekdays_list = ['Monday', 'Tuesday', 'Wednesday']
weekdays_list[0] = 'Mon'  # This works!
print("Modified list:", weekdays_list)

# TUPLE - Fixed (cannot change)
weekdays_tuple = ('Monday', 'Tuesday', 'Wednesday')
# weekdays_tuple[0] = 'Mon'  # This would ERROR!
print("Tuple stays the same:", weekdays_tuple)
````

**Great progress!** You've mastered **Lists** (flexible, changeable) and **Tuples** (fixed, locked).

## 1.3 Sets

**What's the problem Sets solve?**

Imagine you're tracking top 10% students across multiple modules. When you combine these lists, some students appear multiple times - you want **unique values only**!

**Real-world example:**

You're combining top 10% students from Excel, SQL, and Python modules into one master list. Some high performers appear in multiple modules:

```python
combined_top_performers = ['Alice', 'Bob', 'Alice', 'Charlie', 'Bob', 'Diana']
```

You need the **unique list** of all top performers → Use a Set!

**When to use Sets:**

- Remove duplicates from data
- Find common items between groups (intersection)
- Find unique items in one group but not another (difference)
- Check if someone/something is in a collection (fast!)

**What are Sets?**

- Collections that **automatically remove duplicates**
- **Syntax:** Use curly braces `{}` or `set()`
- No specific order (unlike lists)
- Great for checking membership ("Is Alice in the group?")
- Perfect for finding overlaps between groups

````python
# Problem: Combining top 10% lists from Excel, SQL, and Python modules
# Some students are top performers in multiple modules (duplicates!)
combined_list = ['Alice', 'Bob', 'Alice', 'Charlie', 'Bob', 'Diana', 'Alice']

print("Combined list with duplicates:")
print(combined_list)
print(f"Total entries: {len(combined_list)}")

print() #this leaves a blank line for better readability

# Solution: Convert to set - duplicates are automatically removed!
all_top_performers = set(combined_list)

print("Unique top 10% students across all modules (set):")
print(all_top_performers)
print(f"Type: {type(all_top_performers)}")
print(f"Unique students: {len(all_top_performers)}")
````

## 1.3.1 Adding to a Set

Use **`.add()`** to include new items. If the item already exists, it's simply ignored (no error, no duplicate).

Let's add more students who just qualified for top 10% in another module:

````python
# Add a new student who just made top 10% in a module
all_top_performers.add('Eve')
print("After adding Eve:", all_top_performers)

# Try to add someone who's already in the set - nothing happens!
all_top_performers.add('Alice')
print("After trying to add Alice again:", all_top_performers)  # Still no duplicate!
````

## 1.3.2 Removing from a Set

Use **`.discard()`** to remove items. Unlike lists, sets don't have indexes - just remove by value.

Let's remove a student from our master list:

````python
# Remove Bob from the top performers list
all_top_performers.discard('Bob')
print("After removing Bob:", all_top_performers)

# what happens if we try to remove Bob again?
all_top_performers.discard('Bob')  # No error even though Bob is not in the set
````

## 1.3.3 Set Operations

Sets make it easy to compare groups - great for business analysis!



**Union** - Combine all unique elements from sets
- Use **`.union()`** or `|` operator
- Returns **ALL** unique elements from both sets
- **When to use**: Find all students who excelled in at least one module
- Example: Students who scored top 10% in either Excel OR SQL (or both)

**Intersection** - Find common elements between sets
- Use **`.intersection()`** or `&` operator
- Returns elements that appear in **BOTH** sets
- **When to use**: Find students who excelled in both modules (high performers across the board)
- Example: Students who scored top 10% in both Excel AND SQL

**Difference** - Find elements in one set but not another
- Use **`.difference()`** or `-` operator
- Returns elements in first set that are **NOT** in second set
- **When to use**: Find students strong in one module but not another (identify skill gaps)
- Example: Students who scored top 10% in Excel but NOT in SQL

````python
# Now let's compare specific modules: Excel top 10% vs SQL top 10%
# We create fresh, clean sets to clearly demonstrate each operation
Excel_Top10 = {'Alice', 'Bob', 'Charlie', 'Diana'}
SQL_Top10 = {'Charlie', 'Diana', 'Eve', 'Frank'}

# UNION: Find ALL students who excelled in at least one module
print(f"Students who scored top 10% in either Excel OR SQL (Union): \n {Excel_Top10.union(SQL_Top10)}")

# INTERSECTION: Find students who excelled in BOTH modules
print(f"Students who scored top 10% in both Excel AND SQL (Intersection): \n {Excel_Top10.intersection(SQL_Top10)}")

# DIFFERENCE: Find students strong in SQL but not Excel
print(f"Students who scored top 10% in SQL but NOT in Excel (Difference): \n {SQL_Top10.difference(Excel_Top10)}")

# DIFFERENCE: Find students strong in Excel but not SQL
print(f"Students who scored top 10% in Excel but NOT in SQL (Difference): \n {Excel_Top10.difference(SQL_Top10)}")
````

💡 **Why Sets Matter for Data Analytics:**

You'll use Sets constantly in data analysis:
- **Cleaning data**: Remove duplicate customer IDs, transaction records, or survey responses
- **Market analysis**: Find unique products, overlapping customer segments, or exclusive buyers
- **Business reporting**: Identify which customers bought Product A OR Product B, BOTH products, or ONLY one product

Sets make these operations fast and simple - exactly what you need when working with real business data!

## 1.4 Dictionaries

**What's the problem Dictionaries solve?**

Imagine you have a list of phone numbers. With a list, you'd have to remember positions: "Bob's number is at position 2." That's confusing! 

Dictionaries let you use **names as labels** instead: "Look up Bob's number."

**The Problem with Lists:**

```python
# List of phone numbers - which number belongs to whom?
contacts = ['9876-5432', '8765-4321', '6543-3210']  
# Is position 0 Alice or Bob? Hard to remember!
```



**The Solution with Dictionaries:**

```python
# Dictionary with names and numbers - crystal clear!
contacts = {
    'Alice': '9876-5432',
    'Bob': '8765-4321',
    'Charlie': '6543-3210'
}
# Now you can look up Bob's number by his name!
```

**Key things to know about Dictionaries:**

- Store data as **label:value pairs** (we call the label a "key")
- Each key acts like a contact name - it helps you find the right phone number
- Keys must be unique (can't have two people with the exact same name)
- Use `{}` curly braces with colons: `{key: value}`

**How Dictionaries work:**

Think of a dictionary like a phone book:
- **Key** = Person's name (e.g., "Alice", "Bob", "Charlie")  
- **Value** = Their phone number (e.g., "9876-5432")
- **Item** = One complete entry (name + number)

To get a phone number, you look it up by name, not by counting positions!

**Syntax:**
```python
my_dict = {key: value, another_key: another_value}
```

````python
# Create a simple contacts dictionary
contacts = {}  # Start with an empty dictionary

# Add Alice's contact number
#days[0] = 'Monday'
contacts['Alice'] = '+65-1234-5678'
print("After adding Alice:")

print(contacts)
print(f"Number of contacts: {len(contacts)}")
````

````python
# Add more contacts at once with the update method
contacts.update({'Bob': '+65-8765-4321', 'Charlie': '+65-5555-1234', 'Elise' : '+65 8888 8888'})

print("After adding Bob, Elise and Charlie:")
print(contacts)
print()

# Retrieve specific contact information
print("Let us retrieve specific contact information:")
print("Alice's phone number:", contacts['Alice'])
print("Bob's phone number:", contacts['Bob'])
print("Charlie's phone number:", contacts['Charlie'])
````

````python
# View all names (keys)
print(f"All {len(contacts)} contact names: {list(contacts.keys())}")
print()

# View all phone numbers (values)
print(f"All phone numbers: {list(contacts.values())}")

# View complete entries (items)
print(f"Complete contacts information: {list(contacts.items())}")
````

**Retrieving Values with `.get()`**

In addition to using `contacts['key']`, you can also retrieve values using the **`.get()`** method. It's another way to access dictionary values and can be useful in various situations especially in dealing with missing data. While `contacts['key']` works for existing keys, `.get()` helps when data might be incomplete. We will look at an example in a later section [3.1 Understanding `None` - Handling Missing Data](#pnone), where we will better understand how missing values are represented.

````python
# direct access example
print("Alice's phone (direct access):", contacts['Alice'])

# Using .get() - handles missing keys, unlike direct access which causes an error
print("Alice's phone (.get()):", contacts.get('Alice'))
````

**Removing Contacts**

You can remove entries from a dictionary in two ways:
- **`del`** - Simply delete the entry
- **`.pop()`** - Delete the entry AND get back the value

````python
# Remove Alice using del
del contacts['Alice']
print("After removing Alice:")
print(contacts)
````

````python
# Remove Charlie using pop (returns the phone number)
phone = contacts.pop('Charlie')
print(f"Removed Charlie's number: {phone}")

print("Remaining contacts:")
print(contacts)
````

````python
# Update Bob's phone number
contacts['Bob'] = '+65-9999-8888'
print("After updating Bob's number:")
print(contacts)
````

# 2. Type Casting across sets, lists and tuples <a id='p2' />

## Why Convert Between Collection Types?


Sometimes you start with one collection type but need the features of another. For example:

- You have a **set** of unique customer IDs, but need to **sort them** (requires a list)Python makes it easy to convert between collection types using simple functions: `list()`, `set()`, and `tuple()`. Let's see how this works in practice.

- You have a **list** of product codes with duplicates, but need **only unique values** (requires a set)  
- You have data in a **list** that should be **locked** from changes (requires a tuple)

## Converting a Set to a List

**Why?** Sets don't maintain order and can't be sorted. If you need to sort your data or access items by position, convert to a list.

**Business example:** You have unique top performers from different modules, but want to display them alphabetically in a report.

````python
# confirm that all_top_performers is still a set
print(f"As a set: {all_top_performers}")
print(f"Type: {type(all_top_performers)}")

# Convert the set of top performers to a list
top_performers_list = list(all_top_performers)
print(f"As a list: {top_performers_list}")
print(f"Type: {type(top_performers_list)}")

# Now we can sort it!
top_performers_list.sort()
print(f"Sorted: {top_performers_list}")
````

## Converting List to a Set

**Why?** If you receive data as a list or tuple but need to remove duplicates or check for overlaps, convert to a set.

**Business example:** You receive a list with duplicate entries and need to find only unique values.

````python
# Example: list with duplicates
course_registrations = ['Alice', 'Bob', 'Alice', 'Charlie', 'Bob', 'David']
print(f"Original list with duplicates: {course_registrations}")
print(f"Registration count: {len(course_registrations)}")

# Convert to set to get unique students
unique_students = set(course_registrations)
print(f"Unique students (as set): {unique_students}")
print(f"Number of unique students: {len(unique_students)}")
````

## Converting a Set to a Tuple

**Why?** Tuples are locked and can't be changed. Use them when you want to protect important data from accidental modification.

**Business example:** Lock the final list of top performers so it can't be altered after the semester ends.

````python
# Convert to tuple to lock the data
top_performers_tuple = tuple(all_top_performers)
print(f"As a tuple: {top_performers_tuple}")
print(f"Type: {type(top_performers_tuple)}")
````

# 3. Variable-Value Assignments <a id='p3' />

## Working with Multiple Variables at Once

You've been mostly assigning values to single variables like `name = 'Alice'`. Python has convenient shortcuts to work with multiple variables simultaneously - especially useful when unpacking data from collections.

Let's explore the most practical assignment patterns you'll use in data analytics.


````python
# Basic assignment you already know
name = 'Alice'
age = 25
print(f"{name} is {age} years old")
````

## Unpacking Multiple Values

**Why useful?** When data comes in groups (like a row from a database or Excel), you can split it into separate variables in one line.

**Business example:** Splitting employee data (ID, name, department) into individual variables.


````python
# Unpack a tuple into separate variables
employee_data = (12345, 'Sarah Chen', 'Marketing' , 'Director')
emp_id, emp_name, emp_dept, emp_designation = employee_data

print(f"ID: {emp_id}")
print(f"Department: {emp_dept}")
print(f"Name: {emp_name}")
print(f"Desgination: {emp_designation}")
````

## Assigning the Same Value to Multiple Variables

**Why useful?** Initialize several variables with the same starting value.


**Business example:** Setting default values for multiple counters or flags.

````python
# Assign same value to multiple variables
total_sales = revenue = profit = 0
print(f"Starting values - Sales: {total_sales}, Revenue: {revenue}, Profit: {profit}")

# Update one of them
total_sales = 50000
print(f"After update - Sales: {total_sales}, Revenue: {revenue}, Profit: {profit}")
````

## Shorthand for Updating Variables

**Common pattern:** Adding to, subtracting from, or multiplying existing values.

**Business example:** Updating running totals, incrementing counters.


````python
# Long way
sales_total = 100
sales_total = sales_total + 50
print(f"Sales after first update: {sales_total}")

# Shorthand way (better!)
sales_total += 30  # Same as: sales_total = sales_total + 30
print(f"Sales after second update: {sales_total}")

# Works with other operators too
inventory = 200
inventory -= 15  # Subtract
print(f"Inventory after sale: {inventory}")

quantity = 5
quantity *= 3  # Multiply  
print(f"Bulk order quantity: {quantity}")
````

## 3.1 Understanding `None` - Handling Missing Data <a id='pnone'></a>

**What is `None`?**

`None` is Python's way of representing "no value" or "missing data" - like a blank cell in Excel that hasn't been filled in yet. In data analytics, missing values are common (e.g., empty cells in spreadsheets). None helps you represent and handle them without crashing your code.

**Key differences:**
- `None` ≠ zero (`0`) - Zero is a number
- `None` ≠ empty text (`''`) - Empty text is still text
- `None` ≠ `False` - False is a boolean value
- `None` means **nothing is there yet**

**When do we use `None`?**

- As a placeholder when data isn't available yet (like an empty cell in Excel)
- When a function or method doesn't find what you're looking for (e.g., a missing key in a dictionary)
- To signal unknown or missing data in your analysis

**For data analysts:** Think of `None` like a blank cell in your spreadsheet — it means the data is missing, and you need to handle it carefully before doing calculations or analysis.

````python
# Example 1: Using None as a placeholder
customer_email = None  # Email not provided yet
print(f"Email: {customer_email}")

# Example 2: Dictionary lookup with missing keys
contact_info = {'Alice': '+65-1234-5678', 'Bob': '+65-8765-4321'}

# Direct access - will raise KeyError if not found
# Uncommenting the line below will cause an error
# charlie_phone = contact_info['Charlie']

# Safe way to look up - returns None if not found (no error!)
charlie_phone = contact_info.get('Charlie')
print(f"Charlie's phone: {charlie_phone}")  # Shows: None

# You can also provide a default message if the contact is not found
david_phone = contact_info.get('David', 'Not found')
print(f"David's phone: {david_phone}")
````

### **Student Performance Tracker**
Build a program that manages student data using Python collections (lists, tuples, sets, and dictionaries).

**What You Need to Do:**
1. Create a dictionary for a student with keys: "name", "id", "grades" (list of 3 scores), "completed_modules" (list)
2. Create a second student dictionary with similar data but different completed modules
3. Calculate and add the "average_grade" for the first student
4. Use sets to find which modules BOTH students have completed (intersection)
5. Create a tuple with the first student's id and name (to lock this data)
6. Print a formatted report showing student information and common modules

---

**💡 Hint:** Use `set()` on module lists to find intersections. Use `sum()` and `len()` to calculate averages!

````python
# 1. Create the first student dictionary
student1 = {
    "name": "Alice",
    "id": 101,
    "grades": [85, 92, 78],
    "completed_modules": ["Python Basics", "Data Structures", "Loops", "Functions"]
}

# 2. Create the second student dictionary
student2 = {
    "name": "Bob",
    "id": 102,
    "grades": [70, 88, 95],
    "completed_modules": ["Python Basics", "Loops", "OOP", "File Handling"]
}

# 3. Calculate and add average_grade for student1
student1["average_grade"] = sum(student1["grades"]) / len(student1["grades"])

# 4. Use sets to find common completed modules (intersection)
modules1_set = set(student1["completed_modules"])
modules2_set = set(student2["completed_modules"])
common_modules = modules1_set.intersection(modules2_set)
# (equivalent shorthand: common_modules = modules1_set & modules2_set)

# 5. Create a tuple to "lock" student1's id and name
student1_id_locked = (student1["id"], student1["name"])

# 6. Print a formatted report
print("=" * 40)
print("STUDENT PERFORMANCE REPORT")
print("=" * 40)

print(f"\nStudent 1: {student1['name']} (ID: {student1['id']})")
print(f"Grades: {student1['grades']}")
print(f"Average Grade: {student1['average_grade']:.2f}")
print(f"Completed Modules: {student1['completed_modules']}")

print(f"\nStudent 2: {student2['name']} (ID: {student2['id']})")
print(f"Grades: {student2['grades']}")
print(f"Completed Modules: {student2['completed_modules']}")

print(f"\nLocked ID Record (tuple): {student1_id_locked}")

print(f"\nModules Completed by BOTH students: {common_modules}")
print("=" * 40)
````
